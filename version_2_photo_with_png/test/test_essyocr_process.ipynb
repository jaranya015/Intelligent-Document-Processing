```python
import os
import easyocr
import cv2

# 1. ตั้งค่า Path รูปที่ grayscale แล้วใน Version 2
input_folder = '/Users/faiijaran/Desktop/project/version_2/output/photo_by_grayscale'

# 2. โหลดโมเดล EasyOCR (ไทย + อังกฤษ)
print("⌛ กำลังเตรียมโมเดล EasyOCR... (อาจใช้เวลาสักครู่)")
reader = easyocr.Reader(['th', 'en'])

def get_structured_text(image_path):
    # อ่านข้อความจากรูป
    results = reader.readtext(image_path)
    
    # เรียงลำดับพิกัด Y (บนลงล่าง) และ X (ซ้ายไปขวา) เพื่อรักษาโครงสร้างบรรทัด
    results.sort(key=lambda x: (x[0][0][1], x[0][0][0]))
    
    lines = []
    current_line = ""
    last_y = -1
    # ใช้ threshold 10 เพราะไฟล์ดิจิทัลบรรทัดจะตรงและนิ่งมาก
    threshold = 10 

    for (bbox, text, prob) in results:
        top_left_y = bbox[0][1]
        
        if last_y == -1 or abs(top_left_y - last_y) <= threshold:
            current_line += " " + text
        else:
            lines.append(current_line.strip())
            current_line = text
        last_y = top_left_y
        
    if current_line:
        lines.append(current_line.strip())
        
    return "\n".join(lines)

# 3. เริ่มดึงข้อความและแสดงผล
print(f"🚀 เริ่มดึงข้อความจากไฟล์ใน: {input_folder}")
print("=" * 60)

# ดึงรายชื่อไฟล์ gray_1 ถึง gray_10
files = [f for f in os.listdir(input_folder) if f.startswith('gray_') and f.endswith('.png')]
# เรียงลำดับไฟล์ตามตัวเลขเพื่อให้ดูง่าย
files.sort(key=lambda x: int(''.join(filter(str.isdigit, x))))

for filename in files:
    img_path = os.path.join(input_folder, filename)
    
    try:
        # ประมวลผล OCR
        structured_text = get_structured_text(img_path)
        
        # --- แสดงผลลัพธ์ออกที่หน้าจอ Jupyter ---
        print(f"📍 [ผลลัพธ์จากไฟล์: {filename}]")
        print("-" * 30)
        if structured_text.strip():
            print(structured_text)
        else:
            print("(ไม่พบข้อความในรูปนี้)")
        print("=" * 60)
        
    except Exception as e:
        print(f"❌ เกิดข้อผิดพลาดที่ไฟล์ {filename}: {e}")
        print("=" * 60)

print("\n✨ ตรวจสอบข้อมูลในคอนโซลด้านบนได้เลยครับ!")
```

```python
import os
import easyocr
import cv2

# 1. ตั้งค่า Path รูปภาพ
input_folder = '/Users/faiijaran/Desktop/project/version_2/output/photo_by_grayscale'

# 2. โหลดโมเดล
print("⌛ กำลังโหลดโมเดล EasyOCR...")
reader = easyocr.Reader(['th', 'en'])

def get_perfect_lines(image_path):
    # อ่านข้อความ (ปิด paragraph เพื่อให้ได้บล็อกคำที่ละเอียดขึ้น ช่วยเรื่องจัดคอลัมน์)
    results = reader.readtext(image_path, paragraph=False)
    
    if not results:
        return ""

    # ขั้นตอนที่ 1: เรียงลำดับพิกัด Y (บนลงล่าง) เป็นหลัก
    results.sort(key=lambda x: x[0][0][1])
    
    lines = []
    if results:
        current_line_blocks = [results[0]]
        # threshold สำหรับไฟล์ดิจิทัลแนะนำที่ 5-8 พิกเซล เพื่อแยกบรรทัดที่อยู่ใกล้กัน
        y_threshold = 8 

        for i in range(1, len(results)):
            prev_y = current_line_blocks[-1][0][0][1]
            curr_y = results[i][0][0][1]
            
            # ถ้าพิกัด Y ใกล้กันมาก ถือว่าเป็นบรรทัดเดียวกัน
            if abs(curr_y - prev_y) <= y_threshold:
                current_line_blocks.append(results[i])
            else:
                # ก่อนจะจบหน้าบรรทัด ให้เรียงลำดับ X (ซ้ายไปขวา) ภายในบรรทัดนั้นก่อน
                current_line_blocks.sort(key=lambda x: x[0][0][0])
                line_text = " ".join([b[1] for b in current_line_blocks])
                lines.append(line_text)
                
                # เริ่มบรรทัดใหม่
                current_line_blocks = [results[i]]
        
        # เก็บตกบรรทัดสุดท้าย
        current_line_blocks.sort(key=lambda x: x[0][0][0])
        lines.append(" ".join([b[1] for b in current_line_blocks]))

    return "\n".join(lines)

# 3. รันแสดงผล
print(f"🚀 เริ่มดึงข้อมูลแบบเรียงบรรทัด...")
print("=" * 60)

# ดึงไฟล์ที่ต้องการ (เช่น gray_3.png ที่เป็นรูปแมวที่คุณส่งมา)
# หรือเปลี่ยนเป็น loop ทุกไฟล์เหมือนเดิมได้ครับ
target_files = [f for f in os.listdir(input_folder) if f.startswith('gray_') and f.endswith('.png')]
target_files.sort(key=lambda x: int(''.join(filter(str.isdigit, x))))

for filename in target_files:
    img_path = os.path.join(input_folder, filename)
    
    try:
        final_text = get_perfect_lines(img_path)
        print(f"📍 [ไฟล์: {filename}]")
        print("-" * 30)
        print(final_text)
        print("=" * 60)
    except Exception as e:
        print(f"❌ Error ในไฟล์ {filename}: {e}")

print("\n✨ ลองเช็กดูครับว่าการเรียงคอลัมน์ดีขึ้นไหม!")
```
